```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load dataset
df = pd.read_csv("CensusAtSchool.csv")

# Clean missing values for relevant columns
df_clean = df.dropna(subset=["Gender", "Favorite_School_Subject"])

sns.set_theme(style="whitegrid")

# Draw grouped count plot with Female (red) first and Male (blue) second
g = sns.catplot(
    data=df_clean,
    kind="count",
    x="Favorite_School_Subject",
    hue="Gender",
    hue_order=["Female", "Male"],
    palette=["red", "blue"],
    alpha=0.6,
    height=6,
    aspect=2,
)

g.despine(left=True)
g.set_axis_labels("Favorite Subject", "Number of Students")
g.legend.set_title("Gender")

# Rotate subject labels for better readability
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("CensusAtSchool.csv")

# Create a 2-way table (frequency counts)
two_way_table = pd.crosstab(
    index=df["Favorite_School_Subject"],
    columns=df["Gender"],
    margins=True,  # Adds row and column totals
    margins_name="Total",
)

print(two_way_table)
```
